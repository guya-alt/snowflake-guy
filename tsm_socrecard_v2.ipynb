```python
import snowflake.connector
from snowflake.snowpark import Session
import pandas as pd

# Create Snowflake session
session = Session.builder.getOrCreate()
```

```python
# Get all TSM employees from DIM_EMPLOYEE
query = """
SELECT 
    EMPLOYEE_ID,
    EMAIL,
    FIRST_NAME,
    SURNAME,
    DISPLAY_NAME,
    TITLE,
    DEPARTMENT,
    TEAM,
    FINANCE_DEPARTMENT,
    MANAGER,
    SITE,
    ORIGINAL_START_DATE,
    ACTIVE_EFFECTIVE_DATE
FROM PORT_ANALYTICS_PROD.DWH.DIM_EMPLOYEE
WHERE TEAM = 'TSM'
   OR TITLE ILIKE '%TSM%'
   OR TITLE ILIKE '%technical success manager%'
ORDER BY DISPLAY_NAME
"""

# Execute query and load into DataFrame
tsm_employees_df = session.sql(query).to_pandas()

# Display results
print(f"Total TSM Employees: {len(tsm_employees_df)}")
tsm_employees_df
```

```python
# Summary by manager
manager_summary = tsm_employees_df.groupby('MANAGER').size().reset_index(name='COUNT')
print("\nTSM Employees by Manager:")
manager_summary.sort_values('COUNT', ascending=False)
```

```python
# Summary by site
site_summary = tsm_employees_df.groupby('SITE').size().reset_index(name='COUNT')
print("\nTSM Employees by Site:")
site_summary.sort_values('COUNT', ascending=False)
```
